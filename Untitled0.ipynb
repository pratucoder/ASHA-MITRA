#CONTINUE

```python
#continue---> skip the remainning part of the current iterations ,and move to the next iterations
```

```python
for i in range (1,10):
  if i==5:
    break
  print(i)
```

```python
for i in range (1,10):
  if i==5:
    continue
  print(i)
```
